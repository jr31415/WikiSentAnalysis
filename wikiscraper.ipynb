```python
import requests
from bs4 import BeautifulSoup
```

```python
def scrape_article(urls, names, save_location):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/127.0.0.1 Safari/537.36"
    }
    for url, name in zip(urls, names):
        response = requests.get(url, headers=headers)
        html = response.text

        soup = BeautifulSoup(html, "html.parser")
        for h2 in soup.find_all("h2"):
            h2.insert(0, "_____")

        p_list = soup.find_all(["p", "h2"])
        text = ""
        for p in p_list:
            text += p.get_text() + "\n"

        with open(save_location + "/" + name, 'w') as outfile:
            outfile.write(text)
            outfile.close()

urls = ["https://en.wikipedia.org/w/index.php?title=Russia&oldid=1076182140", "https://en.wikipedia.org/w/index.php?title=Russia&oldid=1074740863",\
       "https://en.wikipedia.org/w/index.php?title=Russia&oldid=1073464748", "https://en.wikipedia.org/w/index.php?title=Russia&oldid=1071989967",\
       "https://en.wikipedia.org/w/index.php?title=Russia&oldid=1070618310", "https://en.wikipedia.org/w/index.php?title=Russia&oldid=1069332484"]
names = ["Russia03092022.txt", "Russia03012022.txt", "Russia02222022.txt", "Russia02152022.txt", "Russia02082022.txt", "Russia02012022.txt"]
save_location = "wikipedia articles/russia"

#scrape_article(urls, names, save_location)
```

```python
def month_name_to_num(month):
    """Turns a text string (month) and turns it into a
    a string that is a number representing said month
    """
    match month:
        case "January":
            return "01"
        case "February":
            return "02"
        case "March":
            return "03"
        case "April":
            return "04"
        case "May":
            return "05"
        case "June":
            return "06"
        case "July":
            return "07"
        case "August":
            return "08"
        case "September":
            return "09"
        case "October":
            return "10"
        case "November":
            return "11"
        case "December":
            return "12"

def format_date(date):
    """Takes a string in the format HH:MM Day Month_Name Year and formats it
    as YYYY MM DD HH:MM so that sort() places all function in date order
    """
    date = date.split(" ")
    day = str(date[1])
    if len(day) == 1:
        day = "0" + day
    month = month_name_to_num(date[2])
    year = str(date[3])
    time = str(date[0][:-1])

    return year + " " + month + " " + day + " " + time

def scrape_range(url, article_name, output_folder):
    urls, dates = [], []
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/127.0.0.1 Safari/537.36"
    }
    response = requests.get(url, headers=headers)
    html = response.text

    soup = BeautifulSoup(html, "html.parser")

    links = soup.find_all(class_="mw-changeslist-date")

    
    for link in links:
        urls.append("https://en.wikipedia.org/" + link.get("href"))
        dates.append(link.get_text())

    names = [article_name + "_" + format_date(date) + ".txt" for date in dates]

    scrape_article(urls, names, output_folder)
    
scrape_range("https://en.wikipedia.org/w/index.php?title=Gateshead_International_Stadium&action=history&date-range-to=2025-04-16&tagfilter=&offset=&limit=75", "test", "wikipedia articles/test")
```
